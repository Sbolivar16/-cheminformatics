## **What does Optimus Chem calculate?**

By providing one or multiple **SMILES**, the notebook automatically computes a wide range of molecular properties and filtering rules, including:

- **Basic physicochemical properties**:  
  - Molecular Weight (MW)  
  - LogP (octanol/water partition coefficient)  
  - Topological Polar Surface Area (TPSA)  
  - Number of Hydrogen Bond Donors (HBD)  
  - Number of Hydrogen Bond Acceptors (HBA)  
  - Number of Rotatable Bonds (RotBonds)  

- **Drug-likeness indicators and screening filters**:  
  - Lipinski (Rule of 5)  
  - Veber  
  - Ghose filter  
  - Egan filter  
  - Muegge filter  
  - Rule of 3 (fragment-based screening)  
  - Brenk and PAINS (structural alerts)  
  - Lead-likeness  
  - Other optimization filters (Pfizer 3/75, GSK 4/400, REOS, etc.)  

- **Additional classifications**:  
  - “Is Drug-Like” (meets general drug criteria)  
  - “Is Lead-Like” (meets lead compound criteria)  
  - “Is Fragment-Like” (meets fragment criteria)  

This enables researchers to **assess chemical quality, drug-likeness, and preliminary toxicity risk** before moving on to more advanced molecular design stages.

# 🧪 Optimus Chemical Analyzer – Output Descriptors

When you analyze molecules with **Optimus**, it calculates physicochemical properties and applies several **drug-likeness rules**.  
Below is a detailed explanation of each descriptor and filter.

---

## 📌 Basic Properties

- **SMILES**: linear representation of the molecule.
- **MW (Molecular Weight)**: mass of the molecule.  
  - Recommended: **≤ 500 Da** (Lipinski Rule of 5).
- **LogP**: octanol/water partition coefficient (hydrophobicity).  
  - Acceptable range: **–0.4 to +5.0**.
- **TPSA (Topological Polar Surface Area)**: polar surface area.  
  - Good oral bioavailability: **≤ 140 Å²**.  
  - To cross the blood-brain barrier: **≤ 90 Å²**.
- **HBD (Hydrogen Bond Donors)**: recommended **≤ 5**.
- **HBA (Hydrogen Bond Acceptors)**: recommended **≤ 10**.
- **RotBonds (Rotatable Bonds)**: molecular flexibility.  
  - Good absorption: **≤ 10**.

---

## 📌 Drug-likeness and Scores

- **Drug_Likeness_Score**: overall score combining different filters.  
- **Is_Drug_Like**: whether the molecule passes general drug-likeness rules.  
- **Is_Lead_Like**: evaluates if the molecule is suitable as a **lead compound**.  
- **Is_Fragment_Like**: evaluates if it fits **fragment-based drug discovery** criteria.  

---

## 📌 Filtering Rules

- **Lipinski Rule of 5**  
  - MW ≤ 500, LogP ≤ 5, HBD ≤ 5, HBA ≤ 10.  
  - Used to evaluate **oral drug-likeness**.  

- **Veber Rule**  
  - RotBonds ≤ 10 and TPSA ≤ 140 Å².  
  - Predicts **good oral bioavailability**.  

- **Ghose Filter**  
  - MW between 160–480, LogP –0.4 to 5.6, atom count 40–70.  
  - **Purpose**: to identify **drug-like chemical space** and exclude too small, too large, or unstable molecules.  

- **Egan Filter**  
  - LogP ≤ 5.88, TPSA ≤ 131 Å².  
  - **Purpose**: to predict **oral absorption** based on lipophilicity and polarity.  

- **Muegge Filter**  
  - MW 200–600, LogP –2 to 5, TPSA ≤ 150 Å².  
  - **Purpose**: to detect **non-drug-like compounds**; widely used in virtual screening libraries.  

- **Rule of 3**  
  - MW < 300, LogP ≤ 3, HBD ≤ 3, HBA ≤ 3, RotBonds ≤ 3.  
  - **Purpose**: defines **fragment-like molecules** for **fragment-based drug design**.  

- **CNS MPO (Central Nervous System Multiparameter Optimization)**  
  - Composite score (0–6). Recommended ≥ 4.  
  - **Purpose**: predicts ability to **cross the blood-brain barrier**.  

- **bRo5 (Beyond Rule of 5)**  
  - Compounds violating Lipinski but still bioavailable.  
  - **Purpose**: to detect **macrocycles, peptides, and natural products** that can be orally active despite being outside classical RO5.  

- **Pfizer 3/75 Rule**  
  - If LogP > 3 and TPSA < 75 Å² → higher risk of toxicity.  

- **GSK 4/400 Rule**  
  - MW < 400 and LogP < 4 → higher chance of clinical success.  

- **REOS (Rapid Elimination Of Swill)**  
  - **Purpose**: to filter out unstable or highly reactive compounds.  

- **PAINS (Pan-Assay Interference Compounds)**  
  - **Purpose**: to detect structures that cause false positives in assays.  

- **Brenk Filter**  
  - **Purpose**: to flag toxic, unstable, or reactive functional groups.  

---

## ✅ Practical Guidelines

- **MW ≤ 500**, **LogP –0.4 to 5**, **TPSA ≤ 140**, **HBD ≤ 5**, **HBA ≤ 10**, **RotBonds ≤ 10** → likely good oral bioavailability.  
- **TPSA ≤ 90** → higher probability of CNS penetration.  
- **Lipinski_Pass = True** → good oral drug candidate.  
- **Fails PAINS/Brenk** → treat with caution.  
- **High Drug_Likeness_Score** → more promising compound.
